```python
# programme qui devra calculer la moyenne des 3 UE à partir de notes données en fonction de leur coef
def moyenne_coef():
  #  i = int(input("Combien de notes différentes avez-vous ? ")) # mais int(input fait planter si autre chose que int
    moyenne = [0, 0, 0]
    coef_total = [0, 0, 0]

    for l in range(101,104):
        notes = float(input("Quelle est la moyenne de la ressource R{} ? ".format(l)))
        for ue in range(3):
            coef = float(input("Rentrez le coefficient de R{} dans l'UE{} : ".format(l, ue + 1)))
            moyenne[ue] += notes * coef
            coef_total[ue] += coef

    resultats = []
    for ue in range(3):
        resultats.append(moyenne[ue] / coef_total[ue])
        print("Moyenne de l'UE{} : {}".format(ue + 1, round(resultats[ue], 2)))
    return resultats


(moyenne_coef())
```

```python
from tkinter import *
import pickle

with open('coef_S1','rb') as fichier:
    mon_objet_unpickler = pickle.Unpickler(fichier)
    coefficients_recuperes = mon_objet_unpickler.load()

L_coef1 = coefficients_recuperes["L_coef1"]
L_coef2 = coefficients_recuperes["L_coef2"]
L_coef3 = coefficients_recuperes["L_coef3"]
matiere = [f"R{i}" for i in range(101,116)] + [f"SAE{i}" for i in range(11,17)]

def couleur(val):
    if val >= 10:
        return "green"
    elif val >= 8:
        return "orange"
    else:
        return "red"

def decision_jury(moyennes):
    # décision estimée à partir des moyennes du S1
    nb_valide = 0
    for moy in moyennes:
        if moy >= 10:
            nb_valide += 1
    if nb_valide == 3:
        return "Année validée : passage en BUT 2 autorisé."
    elif nb_valide == 2:
        if min(moyennes) >= 8:
            return "Année non validée mais passage en BUT 2 autorisé : l'UE sous 10 devra être validée en BUT 2."
        else:
            return "Année non validée et passage en BUT 2 refusé : un redoublement peut être proposé par le conseil des enseignants."
    else:
        return "Année non validée : passage en BUT 2 exclu."

def moyenne(L_coef):
    moy = 0
    coef_tot = 0
    for i in range(len(matiere)):
        note = saisies[i].get().replace(",", ".")
        if note != "":
            moy += float(note) * L_coef[i]
            coef_tot += L_coef[i]
    if coef_tot == 0:
        return None
    return round(moy / coef_tot, 2)


fenetreV5 = Tk()
fenetreV5.title("Calc moyennne V5")
saisies = []
for i in range(len(matiere)):
    Label(fenetreV5, text=matiere[i]).grid(row=i, column=0)
    saisie = Entry(fenetreV5)
    saisie.grid(row=i, column=1)
    saisies.append(saisie)

ligne = len(matiere)
label_ue1 = Label(fenetreV5, text="UE1 : ", width=15)
label_ue1.grid(row=ligne + 1, column=1)
label_ue2 = Label(fenetreV5, text="UE2 : ", width=15)
label_ue2.grid(row=ligne + 2, column=1)
label_ue3 = Label(fenetreV5, text="UE3 : ", width=15)
label_ue3.grid(row=ligne + 3, column=1)
label_decision = Label(fenetreV5, text="", wraplength=250)
label_decision.grid(row=ligne + 4, column=0, columnspan=2)


def calc():
    try:
        moy1 = moyenne(L_coef1)
        moy2 = moyenne(L_coef2)
        moy3 = moyenne(L_coef3)
    except ValueError:
        label_decision.config(text="Erreur : une des notes n'est pas un nombre", fg="red")
        return
    if moy1 is None or moy2 is None or moy3 is None:
        label_decision.config(text="Pas assez de notes pour calculer les 3 UE", fg="red")
        return
    label_ue1.config(text=f"UE1 : {moy1}", fg=couleur(moy1))
    label_ue2.config(text=f"UE2 : {moy2}", fg=couleur(moy2))
    label_ue3.config(text=f"UE3 : {moy3}", fg=couleur(moy3))
    label_decision.config(text=decision_jury([moy1, moy2, moy3]), fg="black")

btn = Button(fenetreV5, text="Calculer", command=calc)
btn.grid(row=ligne, column=1)

fenetreV5.mainloop()
```
