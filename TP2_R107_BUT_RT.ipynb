```python
# programme qui devra calculer la moyenne de notes données en fonction de leur coef
def moyenne_coef():
  #  i = int(input("Combien de notes différentes avez-vous ? "))
    moyenne = 0
    coef_total = 0

    for l in range(101,104):
        notes = float(input("Quelle est la moyenne de la ressource R{} ? ".format(l + 1)))
        coef = float(input("Rentrez le coefficient de votre ressource : "))

        moyenne += notes * coef
        coef_total += coef

    return moyenne / coef_total


(moyenne_coef())
```

```python
def moy_UE1():
    liste_coef=[10,10,7,7,5,6,5,4,2,5,5,20,20,7]
    coef_tot=0
    moy = 0 
    valeur = int(input("Indiquez le nombre de notes que vous avez"))
    liste_note = []
    for k in range(valeur):
        liste_note.append(int(input("Indiquez ici vos notes dans un ordre respectif de vos coef")))
    for i in range(valeur):
        moy += liste_note[i] * liste_coef[i]
        coef_tot += liste_coef[i]
    return moy / coef_tot

moy_UE1()
```
